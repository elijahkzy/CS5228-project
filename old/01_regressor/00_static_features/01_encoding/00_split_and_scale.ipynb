# Split, scale and encode the static features without leakage

The earlier scaled files were fit on the whole training set before any validation, so validation data had already influenced the scaler's mean
and standard deviation. The regressor is time based, so the split must be too: a random split would still let the scaler (and the model) see
statistics from months after the ones being validated. The order here is:

1. **Split by time.** Rent approvals from `VALIDATION_START` onward are the validation set; everything earlier is train. Year and month are part
   of the primary keys (Town, Block, Street, Rent Approval Year, Rent Approval Month), so every key group lands on one side by construction.
   The training file ends in 2025-03, so the validation set is the last months of it.
2. **Fit** a `StandardScaler` on the numeric features and a one-hot encoder on the categorical features using the **train split only**.
3. **Transform** train, validation and the test set with those fitted objects.

Feature lists come from `data/config.py`. Missing values follow the earlier convention (numeric -> -1; numeric category codes -> int, -1 for
missing); a constant fill uses no statistics from the data, so it cannot leak. Everything is saved to the `data/` folder next to this notebook.

```python
import json
import os
import sys

import joblib
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler

sys.path.append('/Users/keltonn/Desktop/Projects/housing_price_regressor/data')
import config as cfg

OUT = '/Users/keltonn/Desktop/Projects/housing_price_regressor/00_static_features/01_encoding/data'
KEYS = ['TOWN', 'BLOCK', 'STREET', 'RENT_APPROVAL_YEAR', 'RENT_APPROVAL_MONTH']   # primary keys: rows sharing them are never split apart
VALIDATION_START = '2024-10'      # validation = rent approvals from this month on; the training file ends in 2025-03
CAT_COLS = cfg.cat_col + cfg.cat_col_num      # string categories plus numeric codes that are really categories

assert not [c for c in cfg.id_col + cfg.num_col + CAT_COLS + [cfg.label] if c not in train.columns]
assert not [c for c in cfg.id_col + cfg.num_col + CAT_COLS if c not in test.columns]
print(f'{len(cfg.num_col)} numeric, {len(cfg.cat_col)} string-categorical and {len(cfg.cat_col_num)} numeric-code categorical features')
```

```python
# Missing values follow the convention of the earlier pipeline: numeric -> -1, numeric category codes -> int with -1 for missing.
# A constant fill uses no statistics from the data, so doing it before the split cannot leak anything.
train = train.copy()
test = test.copy()
print('missing numeric values before the fill:', int(train[cfg.num_col].isna().sum().sum()), '(train),', int(test[cfg.num_col].isna().sum().sum()), '(test)')
for df in (train, test):
    df[cfg.num_col] = df[cfg.num_col].fillna(-1)
    for c in cfg.cat_col_num:
        df[c] = df[c].fillna(-1).astype(int)
    df.insert(0, 'ROW_ID', np.arange(len(df)))      # position in the raw file, so every row can be traced back
assert train[cfg.num_col + CAT_COLS].notna().all().all() and test[cfg.num_col + CAT_COLS].notna().all().all()
```

```python
# 1) Time-based split: everything before VALIDATION_START trains, everything from it onward validates.
# Year and month are part of the primary keys, so every (town, block, street, year, month) group necessarily falls on one side.
period = pd.PeriodIndex(year=train['RENT_APPROVAL_YEAR'], month=train['RENT_APPROVAL_MONTH'], freq='M')
is_val = np.asarray(period >= pd.Period(VALIDATION_START, 'M'))
assert is_val.any() and (~is_val).any(), 'VALIDATION_START must fall inside the months of the training file'
tr = train[~is_val].reset_index(drop=True)
va = train[is_val].reset_index(drop=True)

tr_keys, va_keys = set(map(tuple, tr[KEYS].values)), set(map(tuple, va[KEYS].values))
assert not tr_keys & va_keys, 'a key group appears on both sides'
print(f'train       {len(tr):>7,} rows, {len(tr_keys):>7,} key groups, {period[~is_val].min()} to {period[~is_val].max()} ({period[~is_val].nunique()} months)')
print(f'validation  {len(va):>7,} rows, {len(va_keys):>7,} key groups, {period[is_val].min()} to {period[is_val].max()} ({period[is_val].nunique()} months, {len(va) / len(train):.1%} of the rows)')
print(f'test        {len(test):>7,} rows, {test["RENT_APPROVAL_YEAR"].min()}-{test["RENT_APPROVAL_MONTH"][test["RENT_APPROVAL_YEAR"] == test["RENT_APPROVAL_YEAR"].min()].min():02d} onward (all months after the train split)')
```

```python
# 2) Fit the scaler and the encoder on the TRAIN split only; validation and test are only ever transformed
scaler = StandardScaler().fit(tr[cfg.num_col])
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore', dtype=np.int8)
encoder.set_output(transform='pandas')
encoder.fit(tr[CAT_COLS])
print(f'scaler fit on {scaler.n_samples_seen_:,} train rows; encoder produces {len(encoder.get_feature_names_out())} one-hot columns')

# categories in validation/test that the train split never saw: they encode to all zeros
unseen = {}
for name, df in (('validation', va), ('test', test)):
    for col, known in zip(CAT_COLS, encoder.categories_):
        n = int((~df[col].isin(known)).sum())
        if n:
            unseen[(name, col)] = n
print('rows with a category unseen in the train split:', unseen or 'none')
print('constant columns in the train split:', [c for c in cfg.num_col + CAT_COLS if tr[c].nunique() == 1])
```

```python
def process(df, with_label):
    # raw id columns (and the label) are kept as they are; numeric features scaled; categorical features one-hot encoded
    parts = [df[['ROW_ID'] + cfg.id_col]]
    if with_label:
        parts.append(df[[cfg.label]])
    parts.append(pd.DataFrame(scaler.transform(df[cfg.num_col]), columns=cfg.num_col, index=df.index))
    parts.append(encoder.transform(df[CAT_COLS]))
    return pd.concat(parts, axis=1)


train_processed, val_processed, test_processed = process(tr, True), process(va, True), process(test, False)

# The scaler's statistics come from the train split alone: train is exactly mean 0, while validation and test are not
print('numeric features, largest |mean|  train:', round(train_processed[cfg.num_col].mean().abs().max(), 6),
      '| validation:', round(val_processed[cfg.num_col].mean().abs().max(), 4), '| test:', round(test_processed[cfg.num_col].mean().abs().max(), 4))
print('shapes (rows, columns): train', train_processed.shape, '| validation', val_processed.shape, '| test', test_processed.shape)
assert list(train_processed.columns) == list(val_processed.columns) and train_processed.drop(columns=cfg.label).columns.tolist() == test_processed.columns.tolist()
```

```python
os.makedirs(OUT, exist_ok=True)

# processed feature sets (one-hot columns are stored as 0/1 integers to keep the files small)
train_processed.to_csv(f'{OUT}/train_processed.csv', index=False)
val_processed.to_csv(f'{OUT}/validation_processed.csv', index=False)
test_processed.to_csv(f'{OUT}/test_processed.csv', index=False)

# fitted transformers, so the same scaling and encoding can be applied to any new data
joblib.dump(scaler, f'{OUT}/numeric_scaler.joblib')
joblib.dump(encoder, f'{OUT}/categorical_encoder.joblib')

metadata = {
    'split': {
        'type': 'time-based',
        'validation_start': VALIDATION_START,
        'train_months': [str(period[~is_val].min()), str(period[~is_val].max())],
        'validation_months': [str(period[is_val].min()), str(period[is_val].max())],
        'group_keys': KEYS,
        'groups': {'train': len(tr_keys), 'validation': len(va_keys)},
        'rows': {'train': len(tr), 'validation': len(va), 'test': len(test)},
    },
    'missing_values': 'numeric -> -1; numeric category codes -> int, -1 for missing (constant fill, applied before the split)',
    'scaler': 'sklearn StandardScaler fit on the train split only',
    'encoder': 'sklearn OneHotEncoder(handle_unknown="ignore") fit on the train split only; stored as int8',
    'row_id': 'ROW_ID is the row position in the raw train / test feature file',
    'id_columns': ['ROW_ID'] + cfg.id_col,
    'label': cfg.label,
    'numeric_features': cfg.num_col,
    'categorical_features': CAT_COLS,
    'encoded_columns': list(encoder.get_feature_names_out()),
}
with open(f'{OUT}/preprocessing_metadata.json', 'w') as f:
    json.dump(metadata, f, indent=2)

pd.Series({name: f'{os.path.getsize(f"{OUT}/{name}") / 1e6:.1f} MB' for name in sorted(os.listdir(OUT))})
```
